## Initialize MultiPDFExtractor
 
 This code initializes the `MultiPDFExtractor` class, which is responsible for loading PDF files, extracting structured data, and saving vectorized representations in a database.

```python
import sys
sys.path.append("..")

from Extractor import MultiPDFExtractor

def pprint(multiline_string : str):
    for line in multiline_string.splitlines():
        print(line)
```

## Load PDFs from Directory

This cell loads all PDF files from a specified directory and combines pages into single documents.

```python
pdf_dir = "../articles-lite"
extractor = MultiPDFExtractor(pdf_directory=pdf_dir)
```

```python
documents = extractor.load_pdfs()
article1_content = documents[0].page_content
```

```python
article1_content[:1500]
```

## Process Single PDF Document
This cell processes a single PDF document to extract structured data and metadata, chunk it, embed it, and store it in the vector database.

```python
meta_data = extractor.extract_metadata_with_llm(article1_content)
meta_data
```

```python
chunks = extractor.chunk_text(article1_content)
chunks[3:5]
```

```python
embeddings = extractor.embed_chunks(chunks)
embeddings[3:5]
```

```python
extractor.save_to_chroma(chunks, embeddings, meta_data)
```
